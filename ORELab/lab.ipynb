```python
import pandas as pd

market_file = '../Examples/Input/market_20160205.txt'


def read_and_clean_file(file_path):
    """
    Read a text file and remove blank lines and commented lines.
    
    Args:
        file_path: Path to the text file
        
    Returns:
        List of cleaned lines
    """
    cleaned_lines = []
    
    with open(file_path, 'r') as f:
        for line in f:
            # Strip whitespace from both ends
            stripped = line.strip()
            
            # Skip blank lines and lines starting with # (comments)
            if stripped and not stripped.startswith('#'):
                cleaned_lines.append(stripped)
    
    return cleaned_lines

# Read cleaned data to DataFrame
from io import StringIO

cleaned_lines = read_and_clean_file(market_file)
print(f"Total cleaned lines: {len(cleaned_lines)}")

# Join lines and read as CSV (space-separated)
data_str = '\n'.join(cleaned_lines)
df = pd.read_csv(StringIO(data_str), sep='\s+', header=None)
df.columns = ['Date', 'Quote', 'Value']

# Display DataFrame info
print(f"\nDataFrame shape: {df.shape}")
print(f"\nFirst 10 rows:")
df.head(10)
```
